## Preprocessing

### Scaling (neural models perform better when the data are scaled)

So that's a numpy array, all the variables here are numpy arrays. But to use PyTorch Training Pipeline they need to be transferred into PyTorch Tensors.

```python
# transformed_tensor_var = torch.tensor(numpy_array_name)

X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## Defining the Model

```python
X_train_tensor.shape
```

```python
class MySimpleNN():

  def __init__(self, X): # constructor (self(to call w & b), input)
    # initializing weight (size: (30*1)) (to match the matmul operation without Transpose matrix)
    self.weights = torch.rand(X.shape[1], # num of rows = no of input/cols (num of col in input dataset)
                              1, # num of col = 1 # as all features going to only one neuron
                              dtype=torch.float64, # must be float, gradient of int type can't be calculated
                              requires_grad=True)  # keeps track

    # initializing bias (1*1)
    self.bias = torch.zeros(1, # 1 element matrix for the only one neuron
                            dtype=torch.float64,
                            requires_grad=True)

  def forward(self, X): # (self, input); Z = WX + b
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):

    # Clamp predictions to avoid log(0) or log(1) respective to undefined or 0.
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon) # var = torch.clamp(var, min, max)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

You can check the weights and biases after training the pipeline

### Important Parameters (lr and epoch)

```python
learning_rate = 0.1
epochs = 25
```

## Training Pipeline

```python
# create model
model = MySimpleNN(X_train_tensor) # training the model on train dataset

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor)

  # loss calculate
  loss = model.loss_function(y_pred, y_train_tensor)

  # backward pass
  loss.backward()

  # parameters update: so it renew every time ~ = the loop doesnt change the values each time of w_old.
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}') # epoch + 1 as count starts from zero
```

Notice how y_pred is also tracking. We've already seen this in the computational graph. After sigmoid operation we get y_pred. So y_pred is also in requires_grad.

```python
# model evaluation
with torch.no_grad(): # y_pred is also tracking so it will change the value with each loop; that's why the whole thing needs to be under no_grad()
  y_pred = model.forward(X_test_tensor) # getting the y_pred tensor
  y_pred = (y_pred > 0.9).float() # y_pred True if greater than 0.9. Then applying float to convert it to 0/1 (as accuracy range is 0 to 1)
  # accuracy = all true assumptions/ all assumptions
  accuracy = (y_pred == y_test_tensor).float().mean() # accuracy = if y_pred = y (T/F); then converting to float. Then the mean of all the 1 + 0! (! Dividing by all assumptions means mean)
  print(f'Accuracy: {accuracy.item()}')
```
